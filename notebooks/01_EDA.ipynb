```python
# Data Manipulation
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Display Settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

# Plot Style
plt.style.use("ggplot")
```

```python
df = pd.read_csv("../data/Synthetic_Dataset.csv")

print("Dataset Loaded Successfully!")
```

```python
df["Safety_Level"].value_counts()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(
    data=df,
    x="Safety_Level",
    order=df["Safety_Level"].value_counts().index
)

plt.title("Safety Level Distribution")
plt.xlabel("Safety Level")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="Safety_Score",
    bins=25,
    kde=True
)

plt.title("Distribution of Safety Score")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(
    data=df,
    x="Crime_Rate",
    order=["Low", "Medium", "High"]
)

plt.title("Crime Rate Distribution")

plt.show()
```

```python
# Select numerical columns
numerical_df = df.select_dtypes(include=["int64", "float64"])

# Correlation matrix
corr_matrix = numerical_df.corr()

corr_matrix
```

```python

#heat map
plt.figure(figsize=(12,8))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
#Crime Probability vs Safety Score
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Crime_Probability",
    y="Safety_Score",
    alpha=0.4
)

plt.title("Crime Probability vs Safety Score")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Police_Distance_km",
    y="Safety_Score",
    alpha=0.4
)

plt.title("Police Distance vs Safety Score")

plt.show()
```
